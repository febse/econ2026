## 3. Generate Features with Realistic Constraints

Generate 300 observations with fuel_type (0 or 1) and mileage values. Electric cars (fuel_type=1) have lower average mileage than petrol cars (fuel_type=0).

```r
library(tidyverse)
n_obs <- 300

# Generate fuel_type (0 = petrol, 1 = electric)
fuel_type <- rbinom(n_obs, size = 1, prob = 0.4)

# Generate mileage with realistic constraints
# Petrol cars: mean 100,000 miles, sd 30,000
# Electric cars: mean 40,000 miles, sd 15,000 (newer technology)
mileage <- ifelse(fuel_type == 0,
                  rnorm(n_obs, mean = 100000, sd = 30000),
                  rnorm(n_obs, mean = 40000, sd = 15000))

# Ensure mileage is positive
mileage <- pmax(mileage, 0)
```

## 4. Calculate Car Prices

Apply the price model to calculate the price for each car based on its mileage and fuel type.

```r
# Price model: price = 2000 - 0.05 * mileage + 400 * fuel_type
price <- 20000 - 0.05 * mileage + 400 * fuel_type

# Add some random noise to make data more realistic
noise <- rnorm(n_obs, mean = 0, sd = 500)
price <- price + noise
```

## 5. Create and Inspect the Dataset

Create a data frame with the generated data and display summary statistics and basic information about the dataset.

```r
# Create data frame
cars_data <- tibble(
  fuel_type = factor(fuel_type, levels = c(0, 1), 
                     labels = c("Petrol", "Electric")),
  mileage = mileage,
  price = price
)

# Display first few rows
head(cars_data, 10)

# Summary statistics
summary(cars_data)

# Summary by fuel type
cars_data %>%
  group_by(fuel_type) %>%
  summarise(
    n = n(),
    mean_mileage = mean(mileage),
    sd_mileage = sd(mileage),
    mean_price = mean(price),
    sd_price = sd(price)
  )
```

```r
lm(formula = price ~  fuel_type, data = cars_data) |> summary()
```
